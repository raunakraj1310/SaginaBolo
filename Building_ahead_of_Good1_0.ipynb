```python
def initialize_model_and_tokenizer(ckpt_dir, quantization):
    if quantization == "4-bit":
        qconfig = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_compute_dtype=torch.bfloat16,
        )
    elif quantization == "8-bit":
        qconfig = BitsAndBytesConfig(
            load_in_8bit=True,
            bnb_8bit_use_double_quant=True,
            bnb_8bit_compute_dtype=torch.bfloat16,
        )
    else:
        qconfig = None

    tokenizer = AutoTokenizer.from_pretrained(ckpt_dir, trust_remote_code=True)
    model = AutoModelForSeq2SeqLM.from_pretrained(
        ckpt_dir,
        trust_remote_code=True,
        low_cpu_mem_usage=True,
        quantization_config=qconfig,
    )

    if qconfig == None:
        model = model.to(DEVICE)
        if DEVICE == "cuda":
            model.half()

    model.eval()

    return tokenizer, model


def batch_translate(input_sentences, src_lang, tgt_lang, model, tokenizer, ip):
    translations = []
    for i in range(0, len(input_sentences), BATCH_SIZE):
        batch = input_sentences[i : i + BATCH_SIZE]

        # Preprocess the batch and extract entity mappings
        batch = ip.preprocess_batch(batch, src_lang=src_lang, tgt_lang=tgt_lang)

        # Tokenize the batch and generate input encodings
        inputs = tokenizer(
            batch,
            truncation=True,
            padding="longest",
            return_tensors="pt",
            return_attention_mask=True,
        ).to(DEVICE)

        # Generate translations using the model
        with torch.no_grad():
            generated_tokens = model.generate(
                **inputs,
                use_cache=True,
                min_length=0,
                max_length=256,
                num_beams=5,
                num_return_sequences=1,
            )

        # Decode the generated tokens into text
        generated_tokens = tokenizer.batch_decode(
            generated_tokens,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=True,
        )

        # Postprocess the translations, including entity replacement
        translations += ip.postprocess_batch(generated_tokens, lang=tgt_lang)

        del inputs
        torch.cuda.empty_cache()

    return translations
```

```python
def transcribe(audio):
    if audio is None:
        return ""

    # Gradio may provide (sample_rate, numpy_array)
    sr, y = audio

    # Convert stereo -> mono
    if y.ndim > 1:
        y = y.mean(axis=1)

    # Convert to float32
    y = y.astype(np.float32)

    # Avoid division by zero
    if np.max(np.abs(y)) > 0:
        y = y / np.max(np.abs(y))

    # Convert numpy -> torch
    wav = torch.from_numpy(y).unsqueeze(0)

    # Resample to 16 kHz
    if sr != 16000:
        resampler = torchaudio.transforms.Resample(
            orig_freq=sr,
            new_freq=16000
        )
        wav = resampler(wav)

    # Move to GPU
    wav = wav.to(DEVICE)

    # Indic-Conformer RNNT decoding
    with torch.no_grad():
        transcription = model(wav, "hi", "rnnt")

    return transcription
```

```python
def transcribe_wav(wav_path, chunk_seconds=8):

    print("\n🎧 Loading:", wav_path)

    waveform, sr = torchaudio.load(wav_path)

    # Stereo → mono
    if waveform.shape[0] > 1:
        waveform = torch.mean(
            waveform,
            dim=0,
            keepdim=True
        )

    # Resample → 16 kHz
    if sr != 16000:
        resampler = torchaudio.transforms.Resample(
            orig_freq=sr,
            new_freq=16000
        )
        waveform = resampler(waveform)

    sr = 16000

    total_samples = waveform.shape[1]
    total_seconds = total_samples / sr

    print(f"🎵 Total duration: {total_seconds:.2f} seconds")

    chunk_samples = int(chunk_seconds * sr)

    hindi_chunks = []
    english_chunks = []

    # ------------------------------------------------
    # Process every chunk
    # ------------------------------------------------

    for start in range(0, total_samples, chunk_samples):

        end = min(
            start + chunk_samples,
            total_samples
        )

        chunk = waveform[:, start:end]

        print(
            f"\n🧩 Processing "
            f"{start/sr:.1f}s → {end/sr:.1f}s"
        )

        if chunk.shape[1] < int(0.3 * sr):
            continue

        # --------------------------------------------
        # Hindi ASR
        # --------------------------------------------

        hindi_text = model(
            chunk,
            "hi",
            "rnnt"
        )

        hindi_text = str(hindi_text).strip()

        print("🇮🇳 Hindi:", hindi_text)

        # If ASR found nothing, don't translate
        if not hindi_text:
            print("⏭️ No speech detected — skipping translation")
            continue

        hindi_chunks.append(hindi_text)

        # --------------------------------------------
        # Hindi → English
        # --------------------------------------------

        english_text = translate(
            hindi_text,
            src_lang="hin_Deva",
            tgt_lang="eng_Latn"
        )

        # batch_translate may return a list
        if isinstance(english_text, list):
            english_text = english_text[0]

        english_text = str(english_text).strip()

        print("🇬🇧 English:", english_text)

        english_chunks.append(english_text)

    # ------------------------------------------------
    # Combine final results
    # ------------------------------------------------

    final_hindi = " ".join(hindi_chunks)
    final_english = " ".join(english_chunks)

    print("\n==============================")
    print("🇮🇳 FINAL HINDI:")
    print(final_hindi)

    print("\n🇬🇧 FINAL ENGLISH:")
    print(final_english)

    print("==============================")

    return final_hindi, final_english
```

```python
def init_microphone():
    js = Javascript("""
    if (typeof window.micReady === 'undefined') {
        window.micReady = false;
    }
    window.micDebugLog = [];

    function logMic(msg) {
        window.micDebugLog.push(msg);
        console.log("[MIC]", msg);
    }

    async function initMic() {
        try {
            if (window.micStream) {
                window.micReady = true;
                logMic("Already initialized, reusing.");
                return "already_ready|" + window.micDebugLog.join(" || ");
            }

            logMic("Requesting getUserMedia...");
            const stream = await navigator.mediaDevices.getUserMedia({ audio: true });
            logMic("Got stream: " + stream.id);

            const audioContext = new AudioContext();
            logMic("AudioContext state: " + audioContext.state);

            if (audioContext.state === "suspended") {
                await audioContext.resume();
                logMic("After resume(), state: " + audioContext.state);
            }

            const source = audioContext.createMediaStreamSource(stream);
            const analyser = audioContext.createAnalyser();
            analyser.fftSize = 2048;
            source.connect(analyser);

            window.micStream = stream;
            window.micAudioContext = audioContext;
            window.micAnalyser = analyser;
            window.micDataArray = new Uint8Array(analyser.fftSize);
            window.micReady = true;

            logMic("Mic fully initialized.");
            return "ready|" + window.micDebugLog.join(" || ");

        } catch (err) {
            logMic("ERROR: " + err.name + ": " + err.message);
            return "error|" + err.name + ": " + err.message + "|" + window.micDebugLog.join(" || ");
        }
    }
    initMic
    """)
    display(js)
    result = eval_js("initMic()")
    print("🔍 RAW MIC INIT RESULT:", result)

    status = result.split("|")[0]
    print(f"🔍 Mic init status: {status}")

    if status == "error":
        raise RuntimeError(f"Mic init failed in browser: {result}")

    return result


def cleanup_microphone():
    js = Javascript("""
    function cleanupMic() {
        try {
            if (window.micStream) {
                window.micStream.getTracks().forEach(t => t.stop());
            }
            if (window.micAudioContext) {
                window.micAudioContext.close();
            }
            window.micStream = null;
            window.micAudioContext = null;
            window.micAnalyser = null;
            window.micReady = false;
            return "cleaned";
        } catch (err) {
            return "cleanup_error|" + err.message;
        }
    }
    cleanupMic
    """)
    display(js)
    result = eval_js("cleanupMic()")
    print("🔍 Mic cleanup result:", result)
```

```python
def record_from_browser_vad(silence_duration=2.5, calibration_ms=1000, max_seconds=30, margin=0.015):
    js = Javascript("""
    async function recordSegment(silenceDuration, calibrationMs, maxSeconds, margin) {
        const log = [];
        function L(msg) { log.push(msg); console.log("[REC]", msg); }

        try {
            if (!window.micReady) {
                return "error|Mic not initialized|" + log.join(" || ");
            }

            const analyser = window.micAnalyser;
            const dataArray = window.micDataArray;

            function getRMS() {
                analyser.getByteTimeDomainData(dataArray);
                let sum = 0;
                for (let i = 0; i < dataArray.length; i++) {
                    const v = (dataArray[i] - 128) / 128;
                    sum += v * v;
                }
                return Math.sqrt(sum / dataArray.length);
            }

            L("Calibrating for " + calibrationMs + "ms...");
            const calibSamples = [];
            const calibStart = Date.now();
            while (Date.now() - calibStart < calibrationMs) {
                calibSamples.push(getRMS());
                await new Promise(r => setTimeout(r, 50));
            }
            const baseline = calibSamples.reduce((a, b) => a + b, 0) / calibSamples.length;
            const threshold = baseline + margin;
            L("Baseline=" + baseline.toFixed(4) + " Threshold=" + threshold.toFixed(4));

            if (window.micAudioContext.state !== "running") {
                L("WARNING: AudioContext state is " + window.micAudioContext.state + ", not running!");
            }

            const recorder = new MediaRecorder(window.micStream);
            const chunks = [];
            recorder.ondataavailable = e => { if (e.data.size > 0) chunks.push(e.data); };
            recorder.start();
            L("Recorder started, state=" + recorder.state);

            const startTime = Date.now();
            let silenceStart = Date.now();
            let maxRmsObserved = 0;

            return await new Promise((resolve, reject) => {
                function tick() {
                    const rms = getRMS();
                    if (rms > maxRmsObserved) maxRmsObserved = rms;
                    const elapsed = (Date.now() - startTime) / 1000;

                    if (rms > threshold) {
                        silenceStart = null;
                    } else {
                        if (silenceStart === null) silenceStart = Date.now();
                        if ((Date.now() - silenceStart) / 1000 > silenceDuration) {
                            L("Silence " + silenceDuration + "s reached. Stopping. Max RMS seen=" + maxRmsObserved.toFixed(4));
                            recorder.stop();
                            return;
                        }
                    }

                    if (elapsed > maxSeconds) {
                        L("Max seconds (" + maxSeconds + ") reached. Stopping.");
                        recorder.stop();
                        return;
                    }
                    requestAnimationFrame(tick);
                }

                recorder.onstop = () => {
                    L("Recorder stopped. Chunks: " + chunks.length);
                    const blob = new Blob(chunks, { type: recorder.mimeType });
                    const reader = new FileReader();
                    reader.onloadend = () => resolve("ok|" + reader.result + "|" + log.join(" || "));
                    reader.readAsDataURL(blob);
                };

                recorder.onerror = (e) => {
                    L("Recorder error: " + e.error);
                    resolve("error|Recorder error: " + e.error + "|" + log.join(" || "));
                };

                tick();
            });

        } catch (err) {
            L("EXCEPTION: " + err.name + ": " + err.message);
            return "error|" + err.name + ": " + err.message + "|" + log.join(" || ");
        }
    }
    recordSegment
    """)

    display(js)

    result = eval_js(f"recordSegment({silence_duration}, {calibration_ms}, {max_seconds}, {margin})")

    parts = result.split("|", 2)
    status = parts[0]

    if status == "error":
        error_msg = parts[1] if len(parts) > 1 else "unknown"
        log_trail = parts[2] if len(parts) > 2 else ""
        print(f"🔍 RECORDING ERROR: {error_msg}")
        print(f"🔍 JS log trail: {log_trail}")
        raise RuntimeError(f"Recording failed: {error_msg} | trail: {log_trail}")

    data_url = parts[1]
    log_trail = parts[2] if len(parts) > 2 else ""
    print(f"🔍 Recording JS log trail: {log_trail}")

    audio_bytes = base64.b64decode(data_url.split(",")[1])

    webm_path = "/content/current_recording.webm"
    with open(webm_path, "wb") as f:
        f.write(audio_bytes)

    return webm_path
```

```python
import os
import time
import traceback

recording_control = {"active": False}

def continuous_record_and_transcribe():
    print("\n" + "="*50)
    print("🚀 continuous_record_and_transcribe() STARTED")
    print("="*50)

    hindi_history = ""
    english_history = ""

    try:
        print("Step 1: Initializing mic...")
        init_microphone()
        print("Step 1: ✅ Mic ready.")
    except Exception as e:
        print("❌ MIC INIT FAILED:")
        traceback.print_exc()
        yield f"[MIC INIT ERROR: {e}]", "", None, None
        return

    segment_num = 0

    try:
        while recording_control["active"]:
            segment_num += 1
            print(f"\n🎤 --- Segment {segment_num}: listening ---")

            try:
                webm_path = record_from_browser_vad(
                    silence_duration=2.5,
                    calibration_ms=1000,
                    max_seconds=30,
                    margin=0.015
                )
            except Exception as e:
                print(f"❌ SEGMENT {segment_num} RECORDING FAILED:")
                traceback.print_exc()
                yield (hindi_history + f"\n[SEGMENT {segment_num} ERROR: {e}]",
                       english_history, None, None)
                break

            # Safety check: Ensure webm_path is a valid file, not a directory
            if not webm_path or os.path.isdir(webm_path) or not os.path.exists(webm_path):
                print(f"⚠️ Segment {segment_num}: Invalid webm path generated.")
                yield (hindi_history or "Listening...", english_history or "Waiting for speech...", None, None)
                continue

            size = os.path.getsize(webm_path)
            print(f"🔍 Segment {segment_num}: recorded {size} bytes")

            try:
                wav_path = convert_to_wav(webm_path)
                hindi_text, english_text = transcribe_wav(wav_path)
            except Exception as e:
                print(f"❌ SEGMENT {segment_num} TRANSCRIPTION FAILED:")
                traceback.print_exc()
                yield (hindi_history + f"\n[TRANSCRIBE ERROR: {e}]",
                       english_history, None, None)
                continue

            # Double-check that wav_path is an actual file before serving it to Gradio
            safe_wav_path = wav_path if (wav_path and os.path.exists(wav_path) and not os.path.isdir(wav_path)) else None

            if hindi_text.strip():
                hindi_history += hindi_text + "\n"
                english_history += english_text + "\n"

                # ---- TTS step (Silence -> Translation -> TTS) ----
                try:
                    tts_path, tts_duration = speak_text(english_text)
                except Exception as e:
                    print(f"❌ SEGMENT {segment_num} TTS FAILED:")
                    traceback.print_exc()
                    tts_path, tts_duration = None, 0.0

                # Double-check that tts_path is an actual file before serving it to Gradio
                safe_tts_path = tts_path if (tts_path and os.path.exists(tts_path) and not os.path.isdir(tts_path)) else None

                print(f"✅ Segment {segment_num}: yielding update (with TTS)")
                yield hindi_history, english_history, safe_wav_path, safe_tts_path

                # Pause for TTS playback duration so the mic doesn't hear its own output
                if tts_duration > 0:
                    wait_time = tts_duration + 0.5
                    print(f"🔍 Pausing {wait_time:.1f}s for TTS playback before resuming listening...")
                    time.sleep(wait_time)

            else:
                print(f"(Segment {segment_num}: silence, no update, looping safely)")
                # 🎯 THE CRITICAL FIX: Yield to Gradio even on silence so the UI loop stays active
                # and doesn't pass unhandled directory fallback states!
                yield (
                    hindi_history if hindi_history else "Listening...",
                    english_history if english_history else "Waiting for speech...",
                    None,
                    None
                )

        print("🔁 Loop exited — recording_control['active'] is now False")

    finally:
        print("Cleaning up mic...")
        try:
            cleanup_microphone()
        except Exception:
            traceback.print_exc()
        print("🛑 continuous_record_and_transcribe() FULLY STOPPED")
```

```python
def convert_to_wav_named(webm_path, wav_path):
    waveform, sr = torchaudio.load(webm_path)
    if waveform.shape[0] > 1:
        waveform = torch.mean(waveform, dim=0, keepdim=True)
    if sr != 16000:
        resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
        waveform = resampler(waveform)
    torchaudio.save(wav_path, waveform, 16000)
    return wav_path


def dataurl_to_wav(data_url, out_prefix="live_partial"):
    if not data_url or data_url in ("skip", "empty"):
        return None
    audio_bytes = base64.b64decode(data_url.split(",")[1])
    webm_path = f"/content/{out_prefix}.webm"
    with open(webm_path, "wb") as f:
        f.write(audio_bytes)
    try:
        return convert_to_wav_named(webm_path, f"/content/{out_prefix}.wav")
    except Exception:
        # Partial webm blobs can occasionally be malformed mid-flush — skip that poll
        return None


def asr_only(wav_path):
    waveform, sr = torchaudio.load(wav_path)
    if waveform.shape[0] > 1:
        waveform = torch.mean(waveform, dim=0, keepdim=True)
    if sr != 16000:
        resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
        waveform = resampler(waveform)
    if waveform.shape[1] < int(0.3 * 16000):
        return ""
    return str(model(waveform, "hi", "rnnt")).strip()


def transcribe_and_translate(wav_path):
    hindi_text = asr_only(wav_path)
    if not hindi_text:
        return "", ""
    english_text = translate(hindi_text, src_lang="hin_Deva", tgt_lang="eng_Latn")
    if isinstance(english_text, list):
        english_text = english_text[0]
    return hindi_text, str(english_text).strip()
```

```python
recording_control = {"active": False}


def continuous_record_and_transcribe():
    print("\n🚀 STARTED (live mode)")
    hindi_history = ""
    english_history = ""

    try:
        init_microphone()
    except Exception as e:
        traceback.print_exc()
        yield "", f"[MIC ERROR: {e}]", "", None
        return

    segment_num = 0
    poll_interval = 1.3   # seconds between live-partial checks — raise if it lags

    try:
        while recording_control["active"]:
            segment_num += 1
            print(f"\n🎤 --- Segment {segment_num} ---")

            if start_segment_js() != "started":
                print("❌ Could not start segment")
                break

            last_status = None

            # ---- Poll while browser is listening for this segment ----
            while recording_control["active"]:
                info = get_segment_status()
                status = info.get("status")

                if status != last_status:
                    print(f"🔍 Segment {segment_num} status: {info}")
                    last_status = status

                if status == "error":
                    print(f"❌ Segment {segment_num} error: {info.get('error')}")
                    break

                if status == "listening":
                    wav = dataurl_to_wav(get_partial_audio(), out_prefix="live_partial")
                    if wav:
                        try:
                            live_text = asr_only(wav)
                            if live_text:
                                yield live_text, hindi_history, english_history, None
                        except Exception:
                            traceback.print_exc()

                elif status == "done":
                    break

                time.sleep(poll_interval)

            # ---- Segment finished: finalize with translation ----
            final_wav = dataurl_to_wav(get_final_audio(), out_prefix="current_recording")

            if final_wav:
                hindi_text, english_text = transcribe_and_translate(final_wav)
                if hindi_text:
                    hindi_history += hindi_text + "\n"
                    english_history += english_text + "\n"
                    print(f"✅ Segment {segment_num}: finalized")
                    yield "", hindi_history, english_history, final_wav
                else:
                    print(f"(Segment {segment_num}: silence, discarded)")
                    yield "", hindi_history, english_history, None

        print("🔁 Loop exited")

    finally:
        try:
            cleanup_microphone()
        except Exception:
            traceback.print_exc()
        print("🛑 FULLY STOPPED")
```
